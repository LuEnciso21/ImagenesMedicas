## Introducción

La resonancia magnética funcional (fMRI) ha transformado radicalmente la comprensión de la funcionalidad cerebral y sus mecanismos de adaptación. En el ámbito de la rehabilitación, la resonancia magnética funcional se ha convertido en un instrumento indispensable para evaluar la recuperación neurológica y mejorar las estrategias terapéuticas en personas que padecen una variedad de afecciones, como los accidentes cerebrovasculares (ACV), los traumatismo craneoencefálicos (TCE), las lesiones de la médula espinal y los trastornos neurodegenerativos. Su capacidad para identificar las fluctuaciones de la actividad cerebral en tiempo real y mediante metodologías no invasivas la convierte en un biomarcador muy prometedor para la plasticidad cerebral y la eficacia de los esfuerzos de rehabilitación [1].

## Aplicaciones Clave de la fMRI en el Seguimiento en Rehabilitación:

### 1. Evaluación de la Neuroplasticidad y Reorganización Cortical

La fMRI permite una observación detallada de la reorganización cerebral que ocurre en respuesta a una lesión y a la intervención rehabilitadora. Al comparar los patrones de activación cerebral antes y después de la terapia, los investigadores y clínicos pueden identificar las adaptaciones neuronales que subyacen a la mejora funcional. Por ejemplo, en pacientes con ACV, se ha observado que la recuperación motora se asocia con una reorganización de las redes motoras, incluyendo la activación de áreas ipsilesionales (homolaterales a la lesión) o contralaterales, y cambios en la conectividad funcional de estas regiones [ 2].
Estudios utilizando fMRI han demostrado cómo el cerebro puede reclutar áreas adyacentes o remotas para compensar la función perdida. Esta "redistribución" de la actividad cortical es un sello distintivo de la plasticidad post-lesión [3].

### 2. Monitoreo de la Recuperación Funcional y Pronóstico

La fMRI ofrece la posibilidad de predecir los resultados de la rehabilitación y de monitorear el progreso del paciente a lo largo del tiempo.
Al inicio de la rehabilitación, ciertos patrones de activación o conectividad cerebral pueden indicar un mayor potencial de recuperación. Por ejemplo, en pacientes con ACV, la integridad de las vías corticoespinales o la presencia de actividad residual en la corteza motora primaria puede correlacionarse con una mejor recuperación motora [4]. Esta capacidad predictiva es crucial para la estratificación de pacientes y la personalización de los planes de tratamiento.

### 3. Guía y Optimización de Intervenciones Terapéuticas

La fMRI no solo diagnostica y monitorea, sino que también puede informar y optimizar las estrategias de rehabilitación.
**Neurofeedback con fMRI en Tiempo Real (rt-fMRI NF):** Esta es una aplicación innovadora donde los pacientes reciben retroalimentación en tiempo real sobre su propia actividad cerebral. Esto les permite aprender a autorregular la actividad en regiones cerebrales específicas, potenciando la plasticidad y mejorando funciones como el control motor o la atención [5]. 

4. fMRI en rehabilitación asistida por robot
La resonancia magnética funcional (fMRI) se ha integrado con dispositivos robóticos como el **MR-CHIROD**, un robot compatible con entornos de resonancia diseñado para ejercicios de agarre controlados. Esta tecnología permite evaluar en tiempo real la activación cerebral durante el entrenamiento motor de pacientes con accidente cerebrovascular (ACV). 
Estudios han demostrado que, tras varias semanas de rehabilitación asistida por robot, se observa un aumento en la activación de la corteza motora ipsilesional y una reorganización de las conexiones entre áreas motoras, lo que refleja procesos de plasticidad cerebral. Esta técnica permite personalizar la terapia en función de los cambios funcionales observados en el cerebro del paciente [6].
